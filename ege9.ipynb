### Ровно одно из условий

Логическая связка: **исключающее ИЛИ**. Должно выполниться **ровно одно** условие из нескольких: ни ноль, ни два и больше не подходят.

#### №24347 — три условия

[№24347](https://kompege.ru/task?id=24347)

[(А. Ходарин)](https://vk.com/khodarin) Откройте файл электронной таблицы, содержащей в каждой строке шесть натуральных чисел. Определите количество строк таблицы, содержащих числа, для которых выполнено **ровно одно** из условий:

- максимальное число строки не повторяется;
- в строке первое и последнее числа не совпадают ни с минимальным, ни с максимальным числами строки;
- произведение трёх наибольших чисел строки кратно минимальному числу.

В ответе запишите только число.

```python
cnt=0
for line in open('9_1.txt'):
    a = [int(x) for x in line.split()]
    s = sorted(a)
    usl1 = a.count(max(a)) == 1
    usl2 = a[0] != s[0] and a[0] != s[-1] and a[-1] != s[0] and a[-1] != s[-1]
    usl3 = s[-1]*s[-2]*s[-3] % min(a) == 0
    if usl1+usl2+usl3==1:
        cnt+=1
print(cnt)
```

**Ключевые идеи решения:**

- создаем переменные под каждое условие (они будут возвращать `True`/`False`)
- `True` воспринимается также как 1, поэтому `True+True=2`. Суммируем наши `True`/`False`, эта сумма получается означает сколько условий выполнилось

#### №6925 — повторы и средние значения

[№6925](https://kompege.ru/task?id=6925)

[(Д. Статный)](https://vk.com/azinho) Откройте файл электронной таблицы, содержащей в каждой строке шесть неотрицательных целых чисел. Определите количество строк таблицы, содержащих числа, для которых выполнено **только одно** из условий:

- в строке только одно число повторяется дважды, а остальные не повторяются;
- в строке среднее арифметическое чётных чисел отличается от среднего арифметического нечётных чисел более чем на 50.

В ответе запишите только число.

\*Среднее арифметическое для 0 чисел принять равным нулю.

```python
cnt=0
for line in open('9_3.txt'):
    a = [int(x) for x in line.split()]
    p2 = [x for x in a if a.count(x) == 2]
    np = [x for x in a if a.count(x) == 1]
    usl1 = len(p2) == 2 and len(np)==4 #всего 6 чисел в строке

    ch = [x for x in a if x%2==0]
    nch = [x for x in a if x%2==1]
    if ch:
        sr1 = sum(ch)/len(ch)
    else:
        sr1=0 #Среднее арифметическое для 0 чисел принять равным нулю
    if nch:
        sr2 = sum(nch)/len(nch)
    else:
        sr2=0 #Среднее арифметическое для 0 чисел принять равным нулю
    usl2 = abs(sr1 - sr2) > 50 #abs() возвращает модуль, нам нужно чтобы разница у чисел была больше чем 50

    if usl1+usl2==1:
        cnt+=1
print(cnt)
```

- В строке одно число повторяется дважды, остальные различны;
- Выполнены следующие два условия:
    1. Сумма чётных чисел строки больше суммы нечётных;
    2. В строке существуют как чётные числа, так и нечётные числа.

```python
cnt=0
for line in open('9_4.txt'):
    a = [int(x) for x in line.split()]
    p2 = [x for x in a if a.count(x) == 2]
    np = [x for x in a if a.count(x) == 1]
    usl1 = len(p2) == 2 and len(np) == 3
    ch = [x for x in a if x%2==0]
    nch = [x for x in a if x%2==1]
    usl2 = bool(ch) and bool(nch) and sum(ch)>sum(nch) 
    if usl1 + usl2 >=1:
        cnt+=1
print(cnt)
```

- в строке два разных числа повторяются трижды, одно число не повторяется;

- в строке минимальное число встречается два или три раза, остальные числа без повторений;
- сумма минимального и максимального из неповторяющихся чисел больше суммы других неповторяющихся чисел.

В ответе запишите только число.

```python
cnt = 0
for line in open('9_8.txt'):
    a = [int(x) for x in line.split()]
    np = [x for x in a if a.count(x) == 1]
    if (a.count(min(a)) == 2 and len(np)==5) or (a.count(min(a)) == 3 and len(np)==4):
       if (min(np)+max(np)) > sum(np) - (min(np)+max(np)):
            cnt+=1
print(cnt)
```
